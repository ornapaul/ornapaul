```python
import numpy as np
```

```python
# 3.1 1D NumPy Arrays

list = [20, 27, 9, 12, 31, 17, 4, 50, 39, 35]
# Convert to NumPy Array
arr = np.array(list)

#1. Multiply each item by 4 and print.
print(arr * 4)

#2. Find and print mean.
print(np.mean(arr))

#3. Find and print median.
print(np.median(arr))

#4. Find and print standard deviation.
print(np.std(arr))

#5. Get index of the smallest and largest numbers.
smallest_index = np.argmin(arr)
largest_index = np.argmax(arr)
print(f'Index of smallest number: {smallest_index}, index of largest number: {largest_index}')

#6. Print corresponding values (from #5).
print(f'Smallest number: {arr[smallest_index]}, largest number: {arr[largest_index]}')

#7. Print smallest and largest values.
print(f'Smallest number: {np.min(arr)}, largest number: {np.max(arr)}')

#8. Make and print new array with only values greater than 20.
arr2 = arr[arr>20]
print(f'Values greater than 20: {arr2}')
```

```python
#3.2 2D NumPy Arrays

matrix = np.array([
    [5, 12, 17, 9],
    [8, 21, 14, 3],
    [6, 13, 19, 11]
])

#1. Print shape of array.
print(matrix.shape)

#2. Print third column.
print(matrix[:, 2])

#3. Print second row.
print(matrix[1, :])

#4. Multiply all elements by 2 and print.
print(matrix * 2)

#5. Find and print mean of each column.
column_means = np.mean(matrix, axis=0)
print(column_means)

#6. Find and print overall mean.
overall_mean = np.mean(matrix)
print(overall_mean)

#7. Find max value per row and min value per column.
print(f'Max in each row: {np.max(matrix, axis=1)}')
print(f'Min in each column: {np.min(matrix, axis=0)}')

#8. New shape after flattening?
flattened_matrix = matrix.flatten()
print(f'New shape: {flattened_matrix.shape}')

#9. Find flattened index of smallest and largest values.
min_flat_index = np.argmin(flattened_matrix)
max_flat_index = np.argmax(flattened_matrix)
print(f'Flattened index of smallest: {min_flat_index}, flattened index of largest: {max_flat_index}')

#10. Convert flat indices to row-column coordinates.
min_coord = np.unravel_index(min_flat_index, matrix.shape)
max_coord = np.unravel_index(max_flat_index, matrix.shape)
print(f'Min row-column coordinate:", {min_coord}')
print(f'Max row-column coordinate:", {max_coord}')

#11. Create new array with values greater than 15.
greater_than_15 = matrix[matrix > 15]
print(f'Values that are greater than 15:" {greater_than_15}')

#12. Sort each row and print.
sorted_matrix = np.sort(matrix, axis=1)
print(f'Sorted rows: {sorted_matrix}')
```

```python
#4 Prime Star Systems

arr = np.array([
    [2, 3, 5],
    [14, 6, 8],
    [11, 13, 17],
    [7, 10, 13]
])

def is_prime(n):
    '''Function that checks if a number is prime.'''
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

def prime_star_systems(matrix):
    vectorized_is_prime = np.vectorize(is_prime)
    prime_mask = vectorized_is_prime(matrix)
    rows_with_primes = np.any(prime_mask, axis=1)
    return matrix[rows_with_primes]

print(prime_star_systems(arr))
```

```python
#5 Let's Play Checkers!


def checkerboard():
    board = np.zeros((8, 8), dtype=int) #5.1 8x8 array of zeroes
    board[::2, ::2] = 1 #5.2 alternating pattern (odd rows)
    board[1::2, 1::2] = 1 #5.3 alternating pattern (even rows)
    return board

def reverse_checkerboard():
    board = np.zeros((8, 8), dtype=int) #5.4 reversed (starts with 0)
    board[::2, 1::2] = 1 
    board[1::2, ::2] = 1
    return board

print(checkerboard())
print()
print(reverse_checkerboard())
```

```python
#6 The Expanding Universe

def expansion(universe, expansion_rate):
    
    spaces = " " * expansion_rate
   
    return np.char.join(spaces, universe)

universe = np.array(['galaxy', 'clusters'])

print(expansion(universe, 1))
print(expansion(universe, 2))
```

```python
#7 Second-Brightest Star

def secondDimmest(stars):
    #sorts each row in descending order (highest values first)
    sorted_stars = np.sort(stars, axis=1)[:, ::-1]
    
    #extracts the second column (index 1), which is the second-brightest value
    return sorted_stars[:, 1]

np.random.seed(123)
stars = np.random.randint(500, 2000, (5, 5))

print(stars)
print()
print(secondDimmest(stars))
```
