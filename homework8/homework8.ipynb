```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 1. Function to calculate sine wave
def make_sine_wave(x, A, w):
    return A * np.sin(w * x)

# 2. Array from 0 to 2*pi with 1000 points
x = np.linspace(0, 2 * np.pi, 1000)

#3. Parameter arrays
amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]

linestyles = ['-', '--', '-.', ':', '-']

# 4. Create three side-by-side subplots
fig, axes = plt.subplots(1, 3, figsize=(25, 5))

# Subplot 1: Vary only amplitude (frequency = 1)
for i in range(len(amplitude)):
    amp = amplitude[i]
    y = make_sine_wave(x, amp, 1)
    axes[0].plot(x, y, linestyle=linestyles[i], label=f"A={amp}, w=1")

axes[0].set_title("Varying Amplitude")
axes[0].set_xlabel("x")
axes[0].set_ylabel("y")
axes[0].legend()

# Sub plot 2: Vary only frequency (amplitude = 1)
for i in range(len(frequency)):
    freq = frequency[i]
    y = make_sine_wave(x, 1, freq)
    axes[1].plot(x, y, linestyle=linestyles[i], label=f"A=1, w={freq}")

axes[1].set_title("Varying Frequency")
axes[1].set_xlabel("x")
axes[1].set_ylabel("y")
axes[1].legend()

#Subplot 3: Vary amplitude and frequency together
for amp, freq in zip(amplitude, frequency):
    y = make_sine_wave(x, amp, freq)
    axes[2].plot(x, y, label=f"A={amp}, w={freq}")

axes[2].set_title("Varying Amplitude and Frequency")
axes[2].set_xlabel("x")
axes[2].set_ylabel("y")
axes[2].legend()

plt.tight_layout()

# 8. Display the plots
plt.show()
```
